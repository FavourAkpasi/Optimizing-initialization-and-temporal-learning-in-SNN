```python
import warnings
warnings.filterwarnings('ignore')

from edgesnn.data_utils import load_ucr_data
from edgesnn.plotting import plot_class_distribution, plot_mean_std_series
```

```python
DATASET_DIR = "../Data"
datasets = ["ECG5000", "Wafer", "InsectSound"]
```

```python
for dataset in datasets:
    print(f"--- Analyzing {dataset} ---")
    df_train, df_test = load_ucr_data(DATASET_DIR, dataset)

    if df_train is not None and df_test is not None:
        # 1. Plot Class Distribution
        plot_class_distribution(df_train, df_test, dataset)

        # 2. Plot Mean/Std Dev for the Train Set
        # (We analyze the train set to prevent test-set data leakage)
        plot_mean_std_series(df_train, dataset, split_name="Train")
        print("\n")
```
